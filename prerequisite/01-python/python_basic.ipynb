```python
# TASK 1 - Swap two numbers in a list
#
# Write swap(nums, i, j): swap the items at positions i and j
# IN PLACE (the original list must change) and return the list.
#
# Hint: tuple unpacking a, b = b, a is the idiomatic swap.


def swap(nums, i, j):
    nums[i],nums[j] = nums[j],nums[i]
    return nums


nums = [10, 20, 30]
result = swap(nums, 0, 2)
assert result == [30, 20, 10], f"expected [30, 20, 10], got {result}"
assert nums == [30, 20, 10], "swap must modify the list in place"
print("Task 1 passed")
```

```python
# TASK 2 - Dict of word -> id
#
# Write build_vocab(words): return a dict mapping each word to an int id.
#
# Rules:
#   - ids start at 0
#   - ids follow order of FIRST appearance
#   - repeated words keep their first id
#   - unknown-word lookups must not crash (use .get)
#
# Why it matters: this IS how tokenizers build vocabularies.


def build_vocab(words):
    word_dict = {}
    for index,word in enumerate(words):
        if word not in word_dict:
           word_dict[word] = index
    return word_dict



vocab = build_vocab(["i", "love", "cats", "love"])
assert vocab == {"i": 0, "love": 1, "cats": 2}, f"got {vocab}"
assert vocab.get("dogs", -1) == -1, "unknown word must not crash"
print("Task 2 passed")
```

```python
# TASK 3 - Class with __init__ + one method
#
# Write a class Counter:
#   - __init__(self, start=0): store the starting count
#   - increment(self, by=1): add by to the count, return the NEW count
#
# Remember: self is how a method refers to the instance's own data.


class Counter:
    def __init__(self,start= None):
        self.start = start

    def increment(self,by=None):
        if by == None:
            self.start += 1
        else:
            self.start += by

        return self.start


c = Counter(start=10)
assert c.increment() == 11
c2 = Counter(start=10)
assert c2.increment(by=5) == 15
print("Task 3 passed")
```

```python
# TASK 4 - Read a text file, count unique words
#
# Write count_unique_words(path): read the file at `path`,
# return the number of UNIQUE words.
#
# Rules:
#   - words are separated by whitespace
#   - case-insensitive: "The" and "the" are the same word
#   - test file: sample.txt in this folder


def count_unique_words(path):
    with open(path,"r") as file:
        content = file.read().split()
        word_count= {}
        counter  = 0
        for word in content:
            if word not in word_count:
                word_count[word] = 1
                counter += 1
            else:
                word_count[word] += 1
        return counter


n = count_unique_words("sample.txt")
print("unique words:", n)
assert n == 7, f"expected 7, got {n}"
print("Task 4 passed")
```

```python
# TASK 5 - List comprehension
#
# Write squares_of_evens(nums) using ONE list comprehension:
# return the squares of only the even numbers, in order.
#
# Pattern to remember: [expression for item in list if condition]


def squares_of_evens(nums):
    return [x*x for x in nums if x % 2 == 0]


assert squares_of_evens([1, 2, 3, 4, 5, 6]) == [4, 16, 36]
assert squares_of_evens([1, 3, 5]) == []
print("Task 5 passed")
```

```python
# TASK 6 - Slicing
#
# Write first_half(lst): return the FIRST HALF of the list.
# If the list has odd length, the extra element goes to the first half.
# Must work on empty lists too.
#
# Slicing pattern: lst[start:stop]
import math

def first_half(lst):
    n = len(lst)
    return lst[:(n+1)//2]



assert first_half([1, 2, 3, 4]) == [1, 2]
assert first_half([1, 2, 3, 4, 5]) == [1, 2, 3]
assert first_half([]) == []
print("Task 6 passed")
```

```python
# TASK 7 - Sorting with a key
#
# Write sort_by_second(pairs): sort a list of (word, count) tuples
# by count, DESCENDING. Ties: keep original order.
#
# Pattern: sorted(lst, key=...)
#
# Why it matters: this is how you'd find the top tokens in a corpus.


def sort_by_second(pairs):
    return sorted(pairs,key = lambda x: x[1], reverse=True)


pairs = [("the", 5), ("cat", 2), ("a", 3), ("dog", 2)]
assert sort_by_second(pairs) == [("the", 5), ("a", 3), ("cat", 2), ("dog", 2)]
print("Task 7 passed")
```

```python
# TASK 8 - Word frequency (dict counting)
#
# Write word_freq(words): return a dict word -> count.
# Every word from the list must appear as a key, even count 1.
#
# Pattern: check if key exists, then either add or initialize.


def word_freq(words):
    counter = {}
    for word in words:
        counter[word] = counter.get(word,0) + 1
    return counter


assert word_freq(["a", "b", "a"]) == {"a": 2, "b": 1}
assert word_freq([]) == {}
print("Task 8 passed")
```

```python
# TASK 10 - Generator (yield)
#
# Write countdown(n): a GENERATOR that yields n, n-1, ..., 1
# and yields nothing for n <= 0.
#
# Key difference to internalize: a generator produces values
# lazily, one at a time - it does NOT build the whole list in memory.


def countdown(n):
    while n > 0:
        yield n
        n -= 1


assert list(countdown(3)) == [3, 2, 1]
assert list(countdown(0)) == []
print("Task 10 passed")
```

```python
# TASK 13 - String cleaning (mini tokenizer)
#
# Write tokenize(text): split text into clean lowercase words.
#
# Rules:
#   - lowercase everything
#   - strip punctuation . , ! ? from word edges (not inside)
#   - return words in order
#
# Hint: str.strip() removes characters from the EDGES only.


def tokenize(text):
    words = text.lower().split()
    return [word.strip(",!?.") for word in words]


text = "Hello, world! The cat-sat. Isn't it nice?"
assert tokenize(text) == ["hello", "world", "the", "cat-sat", "isn't", "it", "nice"]
print("Task 13 passed")
```
