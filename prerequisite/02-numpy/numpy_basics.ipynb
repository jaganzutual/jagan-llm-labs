```python
# TASK 1 - Create basic arrays
#
# Write make_arrays(): return a dict with five keys:
#   zeros: np.zeros((3, 4))
#   ones: np.ones((2, 3, 4))
#   arange: np.arange(0, 10, 2)
#   random: np.random.randn(3, 3)
#   from_list: np.array([[1, 2], [3, 4]])
#
# Why it matters: every weight matrix in an LLM starts as one of these calls.


import numpy as np


def make_arrays():
    results = {}
    results["zeros"] = np.zeros((3,4))
    results["ones"] = np.ones((2,3,4))
    results["random"] = np.random.randn(3,3)
    results["from_list"] = np.array([[1,2],[3,4]])
    results["arange"] =  np.array([0, 2, 4, 6, 8])

    return results


result = make_arrays()
assert result["zeros"].shape == (3, 4)
assert result["ones"].shape == (2, 3, 4)
assert np.array_equal(result["arange"], np.array([0, 2, 4, 6, 8]))
assert result["random"].shape == (3, 3)
assert np.array_equal(result["from_list"], np.array([[1, 2], [3, 4]]))
print("Task 1 passed")
```

```python
# TASK 2 - Shape inspection
#
# Write describe(x): return dict with shape, dtype, ndim.
# These three show up in EVERY PyTorch error you will ever see.


def describe(x):
    info = {}
    info["shape"] = x.shape
    info["dtype"] = x.dtype
    info["ndim"] = x.ndim

    return info


info = describe(np.array([[1.0, 2.0], [3.0, 4.0]]))
assert info["shape"] == (2, 2)
assert info["dtype"] == np.float64
assert info["ndim"] == 2
print("Task 2 passed")
```

```python
# TASK 3 - Indexing and slicing
#
# Write slice_demo(x): x is (4, 5). Return dict with:
#   row0: first row, shape (5,)
#   col0: first column, shape (4,)
#   top_left: top-left 2x2, shape (2, 2)
#   every_other_row: rows 0 and 2, shape (2, 5)
#
# Pattern: x[i] row, x[:, j] column, x[a:b, c:d] block, x[::2, :] stride.


def slice_demo(x):
    out = {
        "row0" : x[0,:],
        "col0" : x[:,0],
        "top_left": x[:2, :2],
        "every_other_row":  x[::2, :]
    }

    return out


x = np.arange(20).reshape(4, 5)
print(x)
out = slice_demo(x)
assert np.array_equal(out["row0"], np.array([0, 1, 2, 3, 4]))
assert np.array_equal(out["col0"], np.array([0, 5, 10, 15]))
assert np.array_equal(out["top_left"], np.array([[0, 1], [5, 6]]))
assert np.array_equal(out["every_other_row"], np.array([[0, 1, 2, 3, 4], [10, 11, 12, 13, 14]]))
print("Task 3 passed")
```
